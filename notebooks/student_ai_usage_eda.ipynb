## **13. Numerical Range Validation**

* After examining the descriptive statistics, the next step is to validate the numerical columns against their expected ranges. This process helps identify potentially invalid values that may not be detected by missing-value or duplicate checks.
* The GPA fields are expected to follow a 0–4 scale, while the skill retention score is expected to fall between 0 and 100. Tool diversity, perceived AI dependency, and exam anxiety will be checked against their documented scales.
* Weekly generative AI usage and traditional study hours will also be examined for unusually high or potentially invalid values. Unusual observations will be investigated rather than automatically removed, as extreme values may represent genuine behaviour.

```python
expected_ranges = {
    "Pre_Semester_GPA": (0, 4),
    "Post_Semester_GPA": (0, 4),
    "Tool_Diversity": (1, 5),
    "Perceived_AI_Dependency": (1, 10),
    "Anxiety_Level_During_Exams": (1, 10),
    "Skill_Retention_Score": (0, 100),
    "Weekly_GenAI_Hours": (0, 168),
    "Traditional_Study_Hours": (0, 168),
}

range_results = []

for column, (lower, upper) in expected_ranges.items():
    if column in df.columns:
        invalid_count = (
            (df[column] < lower) | (df[column] > upper)
        ).sum()

        range_results.append({
            "Column": column,
            "Expected_Minimum": lower,
            "Expected_Maximum": upper,
            "Observed_Minimum": df[column].min(),
            "Observed_Maximum": df[column].max(),
            "Invalid_Values": int(invalid_count)
        })

range_summary = pd.DataFrame(range_results)
display(range_summary)
```

### **15.2 Creating AI Usage Groups**

To support group-wise analysis, a categorical feature named `AI_Usage_Group` is created using the `Weekly_GenAI_Hours` variable. Students are classified into four groups: Low, Moderate, High, and Very High, based on their weekly generative AI usage.

The groups are defined as follows:

* **Low:** 5 hours or less per week.
* **Moderate:** More than 5 and up to 10 hours per week.
* **High:** More than 10 and up to 20 hours per week.
* **Very High:** More than 20 hours per week.

This derived feature enables comparisons of examination anxiety, academic performance, and other student-related variables across different AI usage levels. These categories are defined for this analysis and do not represent official classifications of AI usage.


```python
# Create AI usage groups based on weekly generative AI usage

bins = [-np.inf, 5, 10, 20, np.inf]
labels = ["Low", "Moderate", "High", "Very High"]

analysis_df["AI_Usage_Group"] = pd.cut(
    analysis_df["Weekly_GenAI_Hours"],
    bins=bins,
    labels=labels,
    right=True,
    include_lowest=True
)

# Check the number of students in each group
display(analysis_df["AI_Usage_Group"].value_counts().reindex(labels))
```

```python
# Verify AI usage groups and check for missing assignments

print("AI_Usage_Group created:", "AI_Usage_Group" in analysis_df.columns)
print("Missing group assignments:", analysis_df["AI_Usage_Group"].isna().sum())

display(
    analysis_df.groupby("AI_Usage_Group", observed=False)
    ["Weekly_GenAI_Hours"]
    .agg(["count", "min", "max", "mean"])
    .round(2)
)
```


# **17. Bivariate Analysis**

### **17.1 Introduction**

* Bivariate analysis is used to examine relationships between two variables and identify patterns, associations, or differences within a dataset. Unlike univariate analysis, which focuses on individual variables, bivariate analysis helps investigate how two variables may be related.
* In this project, bivariate analysis explores relationships between generative AI usage, traditional study habits, academic performance, perceived AI dependency, examination anxiety, skill retention, and burnout risk among students.
* Appropriate visualization techniques are selected according to the types of variables being analyzed. Scatter plots are used to examine relationships between numerical variables, while correlation heatmaps summarize linear relationships among multiple numerical variables. Box plots and grouped summaries may also be used to compare numerical variables across categories.
* The findings are interpreted descriptively. Observed relationships do not necessarily imply causation.


### **17.2 Relationship Between Generative AI Usage and Traditional Study Hours**
#### **Objective**
To investigate the relationship between the number of hours students spend using generative AI each week and the number of hours they spend studying through traditional methods.

#### **Research Question**
Is higher weekly generative AI usage associated with fewer traditional study hours?

#### **Variables**
* `Weekly_GenAI_Hours`
* `Traditional_Study_Hours`

#### **Visualization**
A scatter plot is used to examine the relationship between weekly generative AI usage and traditional study hours. Each point represents an observation in the dataset.


```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=analysis_df,
    x="Weekly_GenAI_Hours",
    y="Traditional_Study_Hours",
    alpha=0.35
)

plt.title("Weekly Generative AI Usage vs. Traditional Study Hours")
plt.xlabel("Weekly Generative AI Usage (Hours)")
plt.ylabel("Traditional Study Hours per Week")

plt.tight_layout()
plt.show()
```

```python
correlation = analysis_df[
    ["Weekly_GenAI_Hours", "Traditional_Study_Hours"]
].corr(method="pearson").iloc[0, 1]

print(f"Pearson correlation coefficient: {correlation:.3f}")
```

### **Interpretation**

* The scatter plot shows the relationship between weekly generative AI usage and traditional study hours among students. The Pearson correlation coefficient is **−0.157**, indicating a very weak negative linear relationship between the two variables.
* This result suggests that higher weekly generative AI usage is slightly associated with fewer traditional study hours. However, the data points are widely distributed, indicating substantial variation in students' study habits across different levels of AI usage.
* The weak correlation does not support the conclusion that increased generative AI usage necessarily leads to reduced traditional studying. Other factors may influence students' study habits, and correlation alone cannot establish causation.

**Conclusion:** Weekly generative AI usage has a very weak negative linear association with traditional study hours in this dataset.

### **18.2 Weekly Generative AI Usage Across Burnout Risk Levels**

* This analysis compares weekly generative AI usage among students classified into Low, Medium, and High burnout-risk categories. A box plot is used to examine the distribution of AI usage within each group, including the median, spread, and potential outliers.
* A grouped summary table is also generated to compare the number of students and descriptive statistics for weekly AI usage across the three categories.
* The objective is to identify whether differences in AI usage patterns are present among students with different burnout-risk levels.


```python
# Compare weekly AI usage across burnout-risk levels

burnout_order = ["Low", "Medium", "High"]

plt.figure(figsize=(9, 6))

sns.boxplot(
    data=analysis_df,
    x="Burnout_Risk_Level",
    y="Weekly_GenAI_Hours",
    order=burnout_order
)

plt.title("Weekly Generative AI Usage Across Burnout Risk Levels")
plt.xlabel("Burnout Risk Level")
plt.ylabel("Weekly Generative AI Usage (Hours)")

plt.tight_layout()
plt.show()
```

```python
# Summarize weekly AI usage for each burnout-risk category

burnout_usage_summary = (
    analysis_df
    .groupby("Burnout_Risk_Level")["Weekly_GenAI_Hours"]
    .agg(
        Count="count",
        Mean="mean",
        Median="median",
        Standard_Deviation="std",
        Minimum="min",
        Maximum="max"
    )
    .reindex(["Low", "Medium", "High"])
    .round(2)
)

display(burnout_usage_summary)
```

### **Interpretation**

* The analysis shows a clear difference in weekly generative AI usage across the three burnout-risk categories. Students in the Low-risk group reported a mean usage of 4.64 hours per week, compared with 7.35 hours in the Medium-risk group and 15.21 hours in the High-risk group.
* The median weekly AI usage also increased across the categories, from 3.41 hours for Low risk to 5.62 hours for Medium risk and 13.25 hours for High risk. This pattern suggests that higher burnout-risk categories are associated with higher typical weekly AI usage in this dataset.
* The High-risk group also showed the greatest variability, with a standard deviation of 10.61 hours. The box plot additionally identified potential high-usage outliers in the Low- and Medium-risk groups. These observations should not automatically be treated as errors.
* Overall, the results indicate an association between weekly generative AI usage and burnout-risk category. However, this descriptive analysis does not establish causation. Further investigation would be needed to determine whether other factors, such as study habits or academic pressure, may contribute to the observed relationship.

# **19. Key Findings**

The exploratory data analysis examined student generative AI usage, academic performance, examination anxiety, perceived AI dependency, skill retention, and burnout risk using a dataset containing 50,000 student records.

The major findings are summarized below:
* **AI usage patterns:** Weekly generative AI usage averaged 8.43 hours, with a median of 5.80 hours. The distribution was right-skewed, indicating that many students reported relatively low usage while a smaller group reported substantially higher usage.
* **Traditional study hours:** Students reported an average of 11.21 hours of traditional study per week. Weekly generative AI usage had a weak negative correlation with traditional study hours (Pearson's r = -0.16).
* **Perceived AI dependency:** Weekly generative AI usage had a strong positive correlation with perceived AI dependency (r = 0.67). Higher reported AI usage was associated with higher perceived dependency scores.
* **Examination anxiety:** Examination anxiety had a positive correlation with weekly AI usage (r = 0.27) and perceived AI dependency (r = 0.31). Group-wise analysis also showed higher median anxiety scores among students in the higher AI usage groups.
* **Academic performance:** Mean GPA increased from 3.15 before the semester to 3.35 after the semester. Pre-semester and post-semester GPA had a strong positive correlation (r = 0.93). However, weekly AI usage had almost no linear correlation with post-semester GPA (r = -0.02). These observations do not establish that AI usage caused academic improvement.
* **Prompt-engineering skill:** Post-semester GPA was broadly similar across Beginner, Intermediate, and Advanced prompt-engineering skill groups. The Advanced group had a slightly higher mean GPA, but the differences were modest.
* **Skill retention:** The Low, Moderate, and High AI usage groups had similar mean skill-retention scores. The Very High usage group had a lower mean score of 70.25, compared with 75.98–77.08 in the other groups.
* **Burnout risk:** Mean weekly AI usage was higher in the High burnout-risk group than in the Medium and Low groups. This indicates an association in the dataset, not proof that greater AI usage leads to burnout.
* **Data quality:** The initial dataset contained 50,000 records and 16 columns, with no missing cells or duplicate rows detected. The numerical range and categorical whitespace checks also identified no issues in the tested fields. However, the unusually large number of post-semester GPA values exactly equal to 4.00 requires further investigation.

Overall, the analysis identified several relationships among generative AI usage, perceived dependency, examination anxiety, skill retention, and burnout risk. These findings are exploratory and should be interpreted in light of the dataset's limitations. Correlation and group comparisons alone cannot establish cause-and-effect relationships.

# **20. Limitations**

Although this exploratory data analysis provides useful insights into student generative AI usage and related academic and behavioral measures, several limitations should be considered.

* **Dataset representativeness:** The analysis covers the available 50,000 records. Without sufficient information about the sampling method and population, the findings cannot automatically be generalized to all students.
* **Causation cannot be established:** Correlations and group-wise comparisons identify associations but cannot determine whether AI usage causes changes in academic performance, anxiety, dependency, skill retention, or burnout risk.
* **Dataset methodology:** The source, collection process, measurement procedures, and reliability of the recorded variables should be verified before drawing strong conclusions.
* **Potentially unusual GPA distribution:** A total of 4,804 records have a post-semester GPA of exactly 4.00, whereas no records have a pre-semester GPA of exactly 4.00. This pattern should be investigated to determine whether it reflects the underlying data or a data-generation or measurement process.
* **Limited analysis of confounding factors:** Other factors, such as prior academic performance, study habits, course difficulty, and individual circumstances, may influence the observed relationships. Their effects are not established by the present analysis.
* **Correlation limitations:** Pearson correlation measures linear relationships and may not fully capture nonlinear patterns. Further analysis using methods such as Spearman correlation could help examine monotonic relationships, particularly for ordinal variables.
* **Exploratory scope:** This project focuses on data inspection, visualization, descriptive statistics, and relationship analysis. The findings have not, by themselves, been validated through causal inference or predictive modeling.

These limitations should be considered when interpreting the results and identifying directions for future research.
